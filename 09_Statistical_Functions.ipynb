```python
import numpy as np

a = np.array([[1,2,3], [4,5,6]])
print(a)
```

```python
print(np.sum(a))    # sum of all elements of an array
```

```python
print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.min(a))
print(np.max(a))
```
