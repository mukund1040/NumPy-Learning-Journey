### Array initialization method

```python
import numpy as np

# zeros array

zero_arr = np.zeros((2,3))
print(zero_arr)
```

```python
# ones array

one_arr = np.ones((1,3))
print(one_arr)
```

```python
# Full array

full_arr = np.full((3,2), 9)
print(full_arr)
```

```python
# Identity Matrix

id_arr = np.eye(3)
print(id_arr)
```

```python
#  empty array
print(np.empty(2))
```

```python
#  evenly spaced array
print(np.arange(2,10,2))
```

```python
# specific number of equally spaced values between a range
print(np.linspace(2,10,4))
```

```python
# random values array - float
r_rr = np.random.rand(3,2)
print(r_rr)
```

```python
# random values array -int
rint_arr = np.random.randint(1,100,(3,3))
print(rint_arr)
```
