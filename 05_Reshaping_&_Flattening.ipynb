```python
import numpy as np

arr1 = np.array([1,2,3,4,5,6])
print(arr1)
```

```python
arr1
```

```python
reshaped = arr1.reshape((6,1))
print(reshaped)
```

```python
reshaped2 = arr1.reshape((2,3))
print(reshaped2)
```

```python
print(reshaped2.flatten())
```
