### **Creating arrays**

```python
import numpy as np
list1 = [1,2,3]
print(list1)
```

```python
arr_list1 = np.array(list1)
print(arr_list1)
```

```python
print(type(arr_list1))
```

```python
print(arr_list1.ndim)
```

```python
list1 = [1,2,3]
list2 = [4,5,6]
arr_list1_2 = np.array([list1, list2])
print(arr_list1_2)
```

```python
print(arr_list1_2.ndim)
```

```python
list1 = [1,2,3]
list2 = [4,5,6]
list3 = [7,8,9]
arr_list1_3 = np.array([[list1, list2, list3]])
print(arr_list1_3)
```

```python
print(arr_list1_3.ndim)
```
